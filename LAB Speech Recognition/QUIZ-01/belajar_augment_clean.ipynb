```python
# Import Library
import torch
import torchaudio
import librosa
import numpy as np
import soundfile as sf
from pydub import AudioSegment
```

```python
# Load data
bundle = torchaudio.pipelines.WAV2VEC2_ASR_BASE_960H
model = bundle.get_model()

audio_file = 'sample_audio.wav'
audio, sr = librosa.load(audio_file, sr=None)   # versi librosa (array + sample rate)
clip = AudioSegment.from_wav(audio_file)        # versi pydub
```

```python
# Export Function
def export_pydub(audio):
    while True:
        file_name = input("input filename: ")
        if file_name.endswith(".wav"):
            break
        else:
            print("must be ends with .wav")
    audio.export(file_name, format="wav")
    print("Audio saved as", file_name)

def export_librosa(audio):
    while True:
        file_name = input("input filename: ")
        if file_name.endswith(".wav"):
            break
        else:
            print("must be ends with .wav")
    sf.write(file_name, np.clip(audio, -1, 1), sr)
    print("Audio saved as", file_name)
```

```python
# Predict Text Function
def predict_text(audio_file):
    # waveform, sr = torchaudio.load(audio_file)
    # # waveform = waveform.mean(dim=0, keepdim=True)

    audio_np, _ = librosa.load(audio_file, sr=bundle.sample_rate, mono=True)
    waveform = torch.from_numpy(audio_np).unsqueeze(0)   # shape (1, waktu)

    if sr != bundle.sample_rate:
        waveform = torchaudio.functional.resample(waveform, sr, bundle.sample_rate)

    with torch.inference_mode():
        emission, _ = model(waveform)

    class GreedyCTCDecoder(torch.nn.Module):
        def __init__(self, labels):
            super().__init__()
            self.labels = labels
            self.blank = 0

        def forward(self, emission: torch.Tensor) -> str:
            indices = torch.argmax(emission, dim=-1)
            indices = torch.unique_consecutive(indices, dim=-1)
            indices = [i for i in indices if i != self.blank]
            return "".join([self.labels[i] for i in indices])

    decoder = GreedyCTCDecoder(bundle.get_labels())
    transcript = decoder(emission[0])
    transcript = transcript.replace("|", " ")
    return transcript
```

```python
# Augmentation Function
def adjvolume(clip):
    while True:
        nilai = float(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    export_pydub(clip + nilai)

def repeat_audio(clip):
    while True:
        nilai = int(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    export_pydub(clip * nilai)

def fade_in(clip):
    while True:
        nilai = int(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    export_pydub(clip.fade_in(nilai))

def fade_out(clip):
    while True:
        nilai = int(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    export_pydub(clip.fade_out(nilai))

def reverse_audio(clip):
    export_pydub(clip.reverse())

def adjpitch():
    while True:
        nilai = float(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    shifted_audio = librosa.effects.pitch_shift(audio, sr=sr, n_steps=nilai)
    export_librosa(shifted_audio)

def adjspeed():
    while True:
        nilai = float(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    speed_audio = librosa.effects.time_stretch(audio, rate=nilai)
    export_librosa(speed_audio)

def white_noise():
    while True:
        nilai = float(input("Input nilai (1 sampai 5): "))
        if not (1 <= nilai <= 5):
            print("harus antara 1 sampai 5")
        else:
            break
    noise = np.random.normal(0, audio.std(), audio.size)
    export_librosa(audio + noise * nilai)

def random_gain():
    nilai = np.random.uniform(0.5, 1.5)
    print(f"random gain x {nilai:.2f}")
    export_librosa(audio * nilai)
```

```python
def mainmenu():
    while True:
        print("1. Audio Augmentation")
        print("2. Predict Text")
        print("3. Exit")

        choice = input(">> ")
        if choice == "1":
            print("Audio Augmentation Menu")
            aug_menu()
        elif choice == "2":
            transcript = predict_text(audio_file)
            print("Hasil Transkrip: ")
            print(transcript)
        elif choice == "3":
            break
        else:
            print("Invalid Input")
    print("Main Menu Closed")

mainmenu()
```
