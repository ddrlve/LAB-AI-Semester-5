```python
import librosa # membaca file dan return ndarray
import torch

# ini buat PC binus
# from transformers import Wav2Vec2ForCTC, Wav2Vec2Tokenizer

# ini buat google colab
from transformers import AutoProcessor, AutoModelForCTC
```

```python
# model = AutoModelForCTC <- Colab
# model = Wav2Vec2ForCTC <- PC Binus
model = AutoModelForCTC.from_pretrained("facebook/wav2vec2-base-960h")

# tokenizer = AutoProcessor <- Colab
# tokenizer = wav2vec2Tokenizer <- PC Binus
tokenizer = AutoProcessor.from_pretrained("facebook/wav2vec2-base-960h")

# librispeech = 960h
```

```python
# Baca File
path = "./sample_audio.flac"
audio, sample_rate = librosa.load(path, sr=16000)

# raw data
input_values = tokenizer(audio, return_tensors="pt").input_values

# logits, output score -> softmax, sigmoid
logits = model(input_values)
```
