## **Section 2: Variables and Data Types**

```python
nama = "Sigma"
umur = 69
pi = 3.14
is_student = True

# alt + shift + arrow up/down 
print(nama, type(nama))
print(umur, type(umur))
print(pi, type(pi))
print(is_student, type(is_student))
```

```python
# Operators
a = 17
b = 3
print(f"a + b = {a + b}")
print(f"a - b = {a - b}")
print(f"a * b = {a * b}") 
print(f"a / b = {a / b}")
print(f"a / b = {a / b:.2f}") # format 2 decimal places
print(f"a // b = {a // b}") # floor division
print(f"a % b = {a % b}") # modulus sisa pembagian
print(f"a ** b = {a ** b}") # exponentiation pangkat
```

```python
# Operator Perbandingan (outputnya boolean True/False)
print(f"a > b = {a > b}")
print(f"a < b = {a < b}")
print(f"a == b = {a == b}")
print(f"a != b = {a != b}")
print(f"a >= b = {a >= b}")
print(f"a <= b = {a <= b}")
```

```python
# and or not
print(f"(a > b) and (b > 0)", (a > b) and (b > 0)) #ini True
```

## **Section 3: Selection and Repetition**

```python
nilai = 100

if nilai > 90:
    grade = "A"
elif nilai > 80:
    grade = "B"
elif nilai > 70:
    grade = "C"
else:
    grade = "D"

print(f"Nilai: {nilai} -> Grade: {grade}")
```

```python
# Ternary Operator -> Else if else di satu line
status = "Lulus" if nilai >= 65 else "Tidak Lulus"
print(f"Nilai = {nilai} -> Status: {status}")
```

```python
for i in range(100):
    print(i, end = " ")

print()

for i in range(2, 25, 2): #pertama mulai dari 2, sampai 25, stepnya 2
    print(i, end = " ")
```

```python
# print semua value di dalam list
kata = ["buy", "me", "a", "cat"]
for k in kata:
    print(k)
```

```python
# print semua value di dalam list
kata = ["BUY", "me", "a", "cat"]
for k in kata:
    print(k.lower(), end = " ") #agar semua huruf menjadi lowercase
```

```python
for idx, k in enumerate(kata):
    print(f"{idx} - {k}")
```

```python
# while loop
n = 5
while n > 0:
    print(n, end = " ")
    n -= 1
```

```python
for i in range(10):
    if i == 3:
        continue #skip ke iterasi selanjutnya
    if i == 8:
        break #keluar dari loop 
    print(i, end = " ")
```

## **Section 5: Built-in Data Structures**

**4 jenis data structure:**
- List = [] -> terurut, valuenya bisa diubah, bisa duplikat
- Tuple = () -> terurut, valuenya tidak bisa diubah (immutable), bisa duplikat
- Set = {} -> tidak terurut, valuenya bisa diubah, tidak bisa duplikat
- Dictionary = {} -> terurut, valuenya bisa diubah, bisa duplikat, key-nya harus unik, key tidak bisa diubah (immutable)

```python
# List
nilai = [50, 75, 65, 100, 87]

# masukin nilai baru ke dalam list
nilai.append(99)

# masukin ke index yang kita mau, misal index 1, value 90
nilai.insert(1, 90) 

#hapus value 50 dari list, angka spesifik
nilai.remove(50) 

# hapus angka terakhir dari list, tidak perlu tau indexnya
nilai.pop()

print(nilai)
print(len(nilai)) #jumlah value di dalam list = 5
print(sorted(nilai)) #urutkan dari kecil ke besar, tapi tidak mengubah list aslinya
```

```python
# Tuple
tuple = (1, 2, 3, 7, 3, 1)
print(tuple)
```

```python
# Set -> biasanya dipake di diagram venn
angka = {1, 1, 2, 3, 3, 5, 5, 5}
print(angka) #hasilnya {1, 2, 3, 5} karena set tidak bisa duplikat
```

```python
# Dictionary
mahasiswa = {
    "nama": "Dian",
    "nim": "123456",
    "umur": 20,
    "nilai": [90, 100, 95]
}
print(mahasiswa["nama"])
mahasiswa["jurusan"] = "Computer Science" #menambahkan key baru ke dictionary
print(mahasiswa)

print()

print(f"Keys: ", list(mahasiswa.keys()))
print(f"Values`: ", list(mahasiswa.values()))
```

```python
# print pake for loop dictionary
for key, value in mahasiswa.items():
    print(f"{key}: {value}")
```

```python
# titik 2 biar sejajar
# print pake for loop dictionary
for key, value in mahasiswa.items():
    print(f"{key:10s}: {value}")
```
