```python
# Input Output, Data Types, Variables

print("Hello, World!")  # Output

name = input("Enter your name: ")  # Input
print("Welcome, ", name) # Output
print(f"Welcome, {name}")  # Output using f-string

age = 25 #Integer data type
isAboveThirty = False #Boolean data type
ipk = 3.75 #Float data type
name = "John Doe" #String data type

x, y = 20, 30 #Multiple variable assignment
```

```python
# Operators

a = 10
b = 5

# Arithmetic Operators
print(a+b) # Addition
print(a-b) # Subtraction
print(a*b) # Multiplication
print(a/b) # Division
print(a//b) # Floor Division
print(a%b) # Modulus
print(a**b) # Exponentiation

# Comparison Operators
print(a==b) # Equal to
print(a!=b) # Not equal to
print(a>b) # Greater than
print(a<b) # Less than
print(a>=b) # Greater than or equal to
print(a<=b) # Less than or equal to

# Logical Operators
print(a>5 and b<10) # Logical AND
print(a>5 or b<10) # Logical OR
print(not(a>5 and b<10)) # Logical NOT
```

```python
# Conditional Statements
if age > 30:
    print("You are above 30 years old.")
else:
    print("You are 30 years old or younger.")
```

```python
# Repetition Statements
# for loop
for i in range(5):
    print(i)
# range: start, stop, step
for i in range(1, 10, 2):
    print(i)
```

```python
# while loop
i = 0
while i < 5:
    print(i)
    i += 1
```

```python
# List: ordered, mutable, allows duplicates
my_list = [1, 2, 3, 4, 5]
print(my_list[0]) # Output: 1

list.append(int(6)) # Add element to the end of the list
list.insert(2, 10) # Insert element at index 2
list.remove(3) # Remove element from the list
list.sort(reverse=True) # Sort the list in ascending order
print(list) # Output: [10, 2, 4, 5, 6]
```

```python
# Tuple: ordered, immutable, allows duplicates
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple) # Output: 1

```

```python
# set: unordered, mutable, no duplicates
my_set = {1, 2, 3, 4, 5}
print(my_set) # Output: {1, 2, 3, 4, 5}

my_set.add(6) # Add element to the set
my_set.remove(3) # Remove element from the set
print(my_set) # Output: {1, 2, 4, 5, 6}
```

```python
# Dictionary: unordered, mutable, key-value pairs
my_dict = {
    "name": "John",
    "age": 30, 
    "city": "New York"

}
print(my_dict["name"]) # Output: John
my_dict["age"] = 31 # Update value
my_dict["country"] = "USA" # Add new key-value pair
print(my_dict) # Output: {'name': 'John', 'age': 31, 'city': 'New York', 'country': 'USA'}

print(my_dict.keys()) # Output: dict_keys(['name', 'age', 'city', 'country']    
print(my_dict.values()) # Output: dict_values(['John', 31, 'New York', 'USA'])
print(my_dict.items()) # Output: dict_items([('name', 'John'), ('age', 31), ('city', 'New York'), ('country', 'USA')])
```

```python
# Arrays: ordered, mutable, allows duplicates
# Arr[start: stop: step]
numbers = [1, 2, 3, 4, 5]
print(numbers[0]) # Output: 1
print(numbers[1:4]) # Output: [2, 3, 4]
print(numbers[:2]) # Output: [1, 2]
print(numbers[2:]) # Output: [3, 4, 5]
print(numbers[::2]) # Output: [1, 3, 5]
print(numbers[::-1]) # Output: [5, 4, 3, 2, 1]
```

```python
import numpy as np

numbers = [1,2,3,4,5]
arr = np.array(numbers)
print(type(numbers)) # Output: <class 'list'>
print(type(arr)) # Output: <class 'numpy.ndarray'>

arr2 = np.array([
    [1, 2, 3], 
    [4, 5, 6]
])

# shape: ngasih tau size array dan dimensi
print(arr.shape) # Output: (5,)
print(arr2.shape) # Output: (2, 3)

# .ndim: ngasih tau dimensi array
print(arr.ndim) # Output: 1
print(arr2.ndim) # Output: 2

# dtype output: data type array
print(arr.dtype) # Output: int64
print(arr2.dtype) # Output: int64

# np.zeros: buat bikin array kosong
print(np.zeros((2, 3))) # Output: [[0. 0. 0.]

# np.empty: buat bikin array kosong tapi isinya random
print(np.empty((2, 3))) # Output: [[0. 0. 0.]
```

```python
# np.arrange: buat bikin array dengan range tertentu
# arrange(start, stop, step)
print(np.arange(0, 10, 2)) # Output: [0 2 4 6 8]

#np.linspace: buat bikin array dengan range tertentu tapi dibagi rata
# linspace(start, stop, num)
print(np.linspace(0, 10, 5)) # Output: [ 0.  2.5  5.   7.5 10. ]

# np.random.rand: buat bikin array random
# rand(d0, d1, ..., dn)
print(np.random.rand(2, 3)) # Output: [[0.5488135  0.71518937 0.60276338]

# np.random.randint: buat bikin array random integer
# randint(low, high=None, size=None, dtype=int)
print(np.random.randint(0, 10, (2, 3))) # Output:

rng = np.random.default_rng()
int_array = rng.integers(low=1, high=10, size=(2, 3))
print(int_array)
```

```python
# np.vstack: buat nambahin array secara vertikal
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([[7, 8, 9], [10, 11, 12]])
arr3 = np.vstack((arr1, arr2))
print(arr3) # Output: [[ 1  2  3]
```

```python
# operasi scalar
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr + 10) # Output: [[11 12 13]
print(arr - 10) # Output: [[-9 -8 -7]
print(arr * 10) # Output: [[10 20 30]
```

```python
# boolean Indexing
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr[arr > 3]) # Output: [4 5 6] 

# and or not
print((arr > 3) & (arr < 6)) # Output: [[False False False]
print((arr > 3) | (arr < 6)) # Output: [[ True  True  True]
print(~(arr > 3)) # Output: [[ True  True  True]
```
